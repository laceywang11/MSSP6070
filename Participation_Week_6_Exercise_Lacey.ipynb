A. Import Files

```python
import pandas as pd
from google.colab import files
uploaded = files.upload()
df = pd.read_csv("Student Performance Data.csv")
df.head()
```

B. Convert to Numeric

```python
for col in ["math score", "reading score", "writing score", "total score"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

C. Transform DataFrame

```python
Transform_df = lambda d:(
    d.assign(
        Average_score = lambda x:x[["math score", "reading score", "writing score"]].mean(axis=1),
        Grade = lambda x:x.apply(
            lambda s: "A" if s["total score"] >= 270
            else "B" if s["total score"] >= 240
            else "C" if s["total score"] >= 210
            else "D", axis=1),
        Percentage = lambda x:(x["total score"].rank(pct=True) * 100)
    )
)
```

D. Print New DataFrame

```python
Transform_df(df)
```
