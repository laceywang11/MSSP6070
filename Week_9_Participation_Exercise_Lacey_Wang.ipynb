# **Image Downloader**

```python
from google.colab import drive
drive.mount('/content/drive')

import requests
from bs4 import BeautifulSoup
import os
```

```python
keyword = input("Enter a photo category (e.g., cats, dogs): ").strip()
url = 'https://www.flickr.com/search/?text=' + keyword
```

```python
res = requests.get(url)
res.raise_for_status()
```

```python
soup = BeautifulSoup(res.text, 'html.parser')
```

```python
images = soup.find_all('img')
print("Found <img> tags:", len(images))
```

```python
save_path = '/content/drive/MyDrive/MSSP 6070/Modules/Week9'
os.makedirs(save_path, exist_ok=True)
```

```python
count = 0
for img in images:
  src = img.get("src")
  if not src:
    continue

  img_url = src
  if src.startswith('//'):
    img_url = 'https:' + src
  elif not src.startswith("http"):
    img_url = 'https://www.flickr.com' + src

  try:
    r = requests.get(img_url, stream=True, timeout=10, allow_redirects=True)
    r.raise_for_status()

    content_type = r.headers.get('content-type')
    if not content_type or not content_type.startswith('image'):
        print("⚠️ Skipped (not an image):", img_url)
        continue

    count = count + 1
    filename = os.path.join(save_path, keyword + str(count) + '.jpg')
    with open(filename, 'wb') as f:
      for chunk in r.iter_content(chunk_size=8192):
        f.write(chunk)

    print("✅ Downloaded:", filename)

  except requests.exceptions.RequestException as e:
    print(f"❌ Error downloading {img_url}: {e}")

print("🎉 All done! Total images downloaded:", count)
```

This code takes a keyword, finds image links from the Flickr search page, and saves the downloaded images into my Google Drive folder. I was mainly stuck at understanding why some image URLs start with // instead of https://, and how to fix them. Now all the dog photos are stored in:/content/drive/MyDrive/MSSP 6070/Modules/Week9
